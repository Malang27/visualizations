# Exercise 1: Pandas plot 1

```python
# Importer les bibliothèques nécessaires
import pandas as pd
import matplotlib.pyplot as plt

# Créer le DataFrame avec les données fournies
df = pd.DataFrame({
    'name':['christopher','marion','maria','mia','clement','randy','remi'],
    'age':[70,30,22,19,45,33,20],
    'gender':['M','F','F','F','M','M','M'],
    'state':['california','dc','california','dc','california','new york','porto'],
    'num_children':[2,0,0,3,8,1,4],
    'num_pets':[5,1,0,5,2,2,3]
})

# Créer le diagramme à barres avec uniquement l'âge
df.plot(
    x='name',         
    y='age',          
    kind='bar',        
    figsize=(10, 6),  
    title='Âge par personne',  
    rot=45,           
    color='blue'    
)


# Afficher le graphique
plt.show()
```

# Exercise 2: Pandas plot 2

```python
# Importer les bibliothèques nécessaires
import pandas as pd
import matplotlib.pyplot as plt

# Créer le DataFrame avec les données fournies
df = pd.DataFrame({
    'name':['christopher','marion','maria','mia','clement','randy','remi'],
    'age':[70,30,22,19,45,33,20],
    'gender':['M','F','F','F','M','M','M'],
    'state':['california','dc','california','dc','california','new york','porto'],
    'num_children':[4,2,1,0,3,1,0],
    'num_pets':[5,1,0,2,2,2,3]
})

# Créer le nuage de points (scatter plot)
df.plot(
    x='age',             
    y='num_children',    
    kind='scatter',      
    figsize=(8, 6),     
    title='Scatter plot: Number of children and age',
    color='red'          
)

# Afficher le graphique
plt.show()
```

# Exercise 3: Matplotlib 1

```python
import matplotlib.pyplot as plt

# Coordonnées des points
x = [1, 4, 5, 6, 7]
y = [2, 6, 3, 6, 3]

# Créer la figure
plt.figure(figsize=(8, 6))

# Tracer la ligne rouge en pointillé-tiret
plt.plot(x, y, color='red', linestyle='dashdot', linewidth=3)

# Ajouter les points bleus
plt.scatter(x, y, color='blue', s=120)

# Définir les limites des axes
plt.xlim(1, 8)
plt.ylim(1, 8)

# Ajouter les titres
plt.title('Display marker')
plt.xlabel('x - axis')
plt.ylabel('y - axis')

# Afficher le graphique
plt.show()
```

# Exercise 4: Matplotlib 2


```python
import matplotlib.pyplot as plt
import numpy as np

left_data = [5, 7, 11, 13, 17]
right_data = [0.1, 0.2, 0.4, 0.8, -1.6]
x_axis = [0.0, 1.0, 2.0, 3.0, 4.0]

# Créer la figure et le premier axe
fig, ax1 = plt.subplots(figsize=(8, 6))

# Tracer les données sur le premier axe (gauche)
ax1.plot(x_axis, left_data, color='black')
ax1.set_xlabel('')  # Pas de label sur l'axe x pour correspondre à l'image
ax1.set_ylabel('left x-axis')
ax1.set_xlim(0, 4)  # Limiter l'axe x

# Créer le deuxième axe (droit)
ax2 = ax1.twinx()

# Tracer les données sur le deuxième axe
ax2.plot(x_axis, right_data, color='red')
ax2.set_ylabel('right y-axis')

# Ajouter le titre
plt.title('Twin axes')

# Afficher le graphique
plt.show()
```

# Exercise 5: Matplotlib subplots

```python
import matplotlib.pyplot as plt

# Créer les sous-graphiques
fig, axs = plt.subplots(2, 3, figsize=(7, 4))

# Ajuster l'espacement
plt.subplots_adjust(hspace=0.5, wspace=0.5)

# Boucle pour personnaliser chaque sous-graphique
for i, ax in enumerate(axs.flat, 1):
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.text(0.5, 0.5, f'(2, 3, {i})', ha='center')
    ax.set_title(f'Title {i}')

plt.show()
```

# Exercise 6: Plotly 1

```python
import numpy as np
import pandas as pd
import plotly.express as px

# Générer les données aléatoires
np.random.seed(42)  # Pour reproductibilité
returns = np.random.randn(50)
price = 100 + np.cumsum(returns)

# Créer le dataframe
dates = pd.date_range(start='2020-09-01', periods=50, freq='B')
df = pd.DataFrame(zip(dates, price), columns=['Date', 'Company_A'])

# Créer le graphique avec Plotly Express
fig = px.line(df, x='Date', y='Company_A', 
              title="Company_1 stock price",
              )

# Mettre à jour le fond et la grille
fig.update_layout(
    xaxis=dict(showgrid=True, gridcolor='rgba(200, 200, 200, 0.5)'),
    yaxis=dict(showgrid=True, gridcolor='rgba(200, 200, 200, 0.5)')
)

# Afficher le graphique
fig.show()
```

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from plotly.offline import init_notebook_mode

# Initialiser le mode notebook si vous êtes dans Jupyter
init_notebook_mode(connected=True)

# Générer les données aléatoires
np.random.seed(42)  # Pour reproductibilité
returns = np.random.randn(50)
price = 100 + np.cumsum(returns)

# Créer le dataframe
dates = pd.date_range(start='2020-09-01', periods=50, freq='B')
df = pd.DataFrame(zip(dates, price), columns=['Date', 'Company_A'])

# Créer le graphique avec Graph Objects
fig = go.Figure()

# Ajouter la ligne
fig.add_trace(
    go.Scatter(
        x=df['Date'], 
        y=df['Company_A'], 
        mode='lines', 
        line=dict(color='royalblue', width=2)
    )
)

# Personnaliser la mise en page
fig.update_layout(
    title="Company_1 stock price",
    xaxis_title="Date",
    yaxis_title="Company_A",
    plot_bgcolor='rgba(240, 240, 250, 0.5)',
    xaxis=dict(showgrid=True, gridcolor='rgba(200, 200, 200, 0.5)'),
    yaxis=dict(showgrid=True, gridcolor='rgba(200, 200, 200, 0.5)')
)

# Afficher le graphique
fig.show()
```

# Exercice 7 : Box plots de Plotly

```python
import numpy as np
import plotly.express as px
import pandas as pd

# Générer les données aléatoires
np.random.seed(42)  # Pour reproductibilité
y0 = np.random.randn(50)
y1 = np.random.randn(50) + 1  # décaler la moyenne
y2 = np.random.randn(50) + 2

# Préparer les données au format long pour Plotly Express
df = pd.DataFrame({'Sample A': y0,'Sample B': y1,'Sample C': y2})

# Convertir en format long
df_long = pd.melt(df, var_name='Sample', value_name='Value')

# Créer le box plot
fig = px.box(df_long, x='Sample', y='Value', color='Sample')


# Afficher le graphique
fig.show()
```
